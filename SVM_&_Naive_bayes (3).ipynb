#1.: What is a Support Vector Machine (SVM)?


Answer


A Support Vector Machine (SVM) is a powerful and versatile machine learning algorithm used for both classification and regression tasks.

In classification, an SVM works by finding the optimal hyperplane that separates data points belonging to different classes. The goal is to maximize the margin, which is the distance between the hyperplane and the nearest data points from each class (called support vectors). A larger margin generally leads to better generalization performance on unseen data.

For regression, SVMs find a function that approximates the relationship between the input features and the target variable, while keeping the errors within a certain threshold



#: 2. What is the difference between Hard Margin and Soft Margin SVM?


Answer         

       
 The difference between Hard Margin and Soft Margin SVM lies in how they handle misclassifications during training.

Hard Margin SVM:


Aims to find a hyperplane that perfectly separates the data points into their respective classes.
Requires the data to be linearly separable (meaning a single straight line or hyperplane can separate the classes).
Does not allow any data points to be on the wrong side of the margin or the hyperplane.
Can be very sensitive to outliers. If even a single outlier exists, a hard margin SVM may not be able to find a separating hyperplane, or it might find one that generalizes poorly.




Soft Margin SVM:


   
Allows for some misclassifications (or points within the margin) during training.
Introduces a slack variable for each data point, which measures how much a point violates the margin or is on the wrong side of the hyperplane.
Includes a regularization parameter (C) that controls the trade-off between maximizing the margin and minimizing the misclassification errors.
A small C emphasizes a larger margin, potentially allowing more misclassifications.
A large C emphasizes minimizing misclassifications, potentially resulting in a smaller margin.
Is more robust to outliers and can work with data that is not perfectly linearly separable.
In summary, Hard Margin SVM is ideal for perfectly separable data and is sensitive to outliers, while Soft Margin SVM is more flexible, allowing for some errors to handle non-linearly separable data and outliers by introducing a cost for misclassification. In practice, Soft Margin SVM is more commonly used because real-world data is rarely perfectly linearly separable.

# 3. What is the mathematical intuition behind SVM?

Answer    
   The mathematical intuition behind SVM revolves around finding the best possible hyperplane to separate data points into different classes. Let's break it down:


   Hyperplane: In a 2D space, a hyperplane is simply a line. In 3D, it's a plane. In higher dimensions, it's a "flat" subspace that divides the space into two halves. The goal of SVM is to find the hyperplane that best separates the data points of different classes.



Margin: The margin is the distance between the hyperplane and the nearest data points from each class. These nearest data points are called support vectors. The key idea in SVM is to find the hyperplane that maximizes this margin. Why maximize the margin? Because a larger margin generally leads to better generalization performance on unseen data. It creates a "buffer zone" around the decision boundary, making the classification more robust.


Optimization: Finding the optimal hyperplane involves solving an optimization problem. Mathematically, this means finding the weights (W) and bias (b) of the hyperplane equation (W*x + b = 0) that maximizes the margin while ensuring that all data points are on the correct side of the margin (for hard margin SVM) or minimizing the penalty for misclassifications (for soft margin SVM).


Support Vectors: As mentioned, support vectors are the data points closest to the hyperplane. They are crucial because they are the only points that influence the position and orientation of the optimal hyperplane. All other data points can be removed without affecting the decision boundary. This is why SVMs are considered "sparse" models.


Kernel Trick: For non-linearly separable data, SVM uses the "kernel trick." This involves mapping the data into a higher-dimensional space where it might become linearly separable. The kernel function allows us to calculate the dot product of the data points in this higher-dimensional space without explicitly transforming the data. This is a powerful technique that allows SVM to find complex non-linear decision boundaries.







  In essence, SVM's mathematical intuition is about finding the widest possible "street" (margin) between the different classes in the data space, and the support vectors are the points that define the edges of this street. The optimization process finds the parameters that define this "street."




#4.: What is the role of Lagrange Multipliers in SVM?



Answer
 Lagrange multipliers play a crucial role in solving the constrained optimization problem that arises when trying to find the optimal hyperplane in SVM.


The Problem: The goal in SVM is to maximize the margin (or minimize a term related to the weights of the hyperplane, which is equivalent) subject to constraints. These constraints ensure that all data points are on the correct side of the margin (for hard margin SVM) or that the misclassification penalty is within a certain limit (for soft margin SVM).

Constrained Optimization: This type of problem, where you want to optimize a function subject to constraints, is a classic scenario for using Lagrange multipliers. The method involves transforming the constrained optimization problem into an unconstrained one by introducing Lagrange multipliers.

The Lagrangian Function: A new function, called the Lagrangian, is constructed. It combines the objective function (what you want to optimize) with the constraints, each multiplied by a corresponding Lagrange multiplier.


Finding the Saddle Point: The optimization problem then becomes finding the saddle point of the Lagrangian function. This involves taking partial derivatives of the Lagrangian with respect to the weights, bias, and Lagrange multipliers, and setting them to zero.


The Dual Problem: Solving the original (primal) optimization problem directly can be difficult, especially in high-dimensional spaces. By using Lagrange multipliers, the problem can be transformed into its dual form. The dual problem is often easier to solve, particularly when using the kernel trick.


Connection to Support Vectors: A key outcome of the dual problem formulation is that the Lagrange multipliers associated with the constraints will be non-zero only for the support vectors. This elegantly shows why only the support vectors influence the decision boundary. The data points that are not support vectors have a Lagrange multiplier of zero, meaning they don't affect the optimization.



In essence, Lagrange multipliers provide a mathematical framework to incorporate the constraints into the SVM optimization problem, allowing us to find the optimal hyperplane by solving the dual problem, which is often more tractable and highlights the importance of the support vectors.

       
#5: What are Support Vectors in SVM?

Answer  
  Support vectors are a fundamental concept in Support Vector Machines (SVMs)!

In the context of SVM, support vectors are the data points that are closest to the decision boundary (the hyperplane). They are the most crucial data points in determining the position and orientation of the optimal hyperplane that separates the different classes.

Here's why they are so important:

Defining the Margin: The distance between the hyperplane and the support vectors from each class defines the margin. SVM aims to maximize this margin, and the support vectors are the points that lie on the edges of this maximized margin.



Influencing the Hyperplane: Only the support vectors influence the final position of the decision boundary. If you were to remove any data points that are not support vectors, the optimal hyperplane would not change. This is why SVM is sometimes called a "sparse" model.

Handling Non-Linearity (with the Kernel Trick): In cases where the data is not linearly separable in the original feature space, SVM uses the kernel trick to implicitly map the data into a higher-dimensional space where it might become linearly separable. Even in this higher-dimensional space, the support vectors are the points closest to the separating hyperplane in that transformed space.


Computational Efficiency: Because only the support vectors are needed to define the decision boundary, SVM can be computationally efficient during the prediction phase, as you only need to calculate the distance of a new data point to the support vectors.

Think of it like this: If you're trying to draw a line to separate two groups of points, the points that are right on the border, or the closest ones to where you'd draw the line, are the ones that matter most in deciding exactly where that line should go. Those are the support vectors.

# 7. What is a Support Vector Regressor (SVR)?


Answer  
A Support Vector Regressor (SVR) is the application of the Support Vector Machine (SVM) algorithm specifically for regression tasks.

While SVM is often associated with classification, it can also be adapted for regression problems, where the goal is to predict a continuous output value rather than a categorical class label.

Here's how SVR works:

Finding a function: Instead of finding a hyperplane that separates classes, SVR finds a function that approximates the relationship between the input features and the target variable.


Epsilon-Tube: SVR introduces the concept of an "epsilon-tube" around the fitted function. The goal is to find a function that has at most epsilon error from the actual target values for all training data, while still being as flat as possible (to avoid overfitting).

Support Vectors: Similar to SVC, SVR also has support vectors. These are the data points that lie outside the epsilon-tube or on its boundary. These are the points that influence the shape of the fitted function.

Regularization: SVR also uses a regularization parameter (C) to control the trade-off between the flatness of the function and the amount of error allowed

In essence, SVR tries to find a function that fits the data within a certain
 margin of error (defined by epsilon) and uses support vectors to determine the shape of this function. It's a powerful tool for non-linear regression problems, especially when the number of features is large.





#8. What is the Kernel Trick in SVM?


Answer
   
  the Kernel Trick is one of the most elegant and powerful ideas in SVM!

Essentially, the Kernel Trick is a way to handle non-linearly separable data in SVM without explicitly transforming the data into a higher-dimensional space.

Here's the intuition:

The Problem: Sometimes, your data points for different classes aren't neatly separated by a straight line (in 2D) or a flat plane (in 3D). They might be intertwined in a way that requires a curved or more complex boundary.
The Idea: If you could map your data into a higher-dimensional space, it's possible that the data points might become linearly separable in that new space. Then, you could use a standard linear SVM in that higher dimension.
The Challenge: Explicitly transforming the data into a very high-dimensional space can be computationally very expensive, especially if that space is infinite-dimensional.
The Kernel Trick Solution: The Kernel Trick bypasses the need to explicitly map the data. Instead, it uses a kernel function that calculates the dot product of the data points as if they were already in that higher-dimensional space.
The key insight is that the SVM optimization problem, when formulated in its dual form using Lagrange multipliers, only depends on the dot products of the data points. By replacing the standard dot product with a kernel function, you can effectively work in a higher-dimensional space without ever having to represent the data in that space.

Common kernel functions include:

Linear Kernel: This is the simplest kernel, equivalent to the standard dot product, and is used for linearly separable data.
Polynomial Kernel: This maps the data into a higher dimension by using polynomial combinations of the original features.
Radial Basis Function (RBF) Kernel: This is a very popular kernel that can map data into an infinite-dimensional space and is effective for complex non-linear relationships.
In short, the Kernel Trick is a clever mathematical shortcut that allows SVM to find non-linear decision boundaries efficiently by operating in a potentially much higher-dimensional space implicitly through the use of kernel functions




#9. Compare Linear Kernel, Polynomial Kernel, and RBF Kernel:

Answer
 Comparing the Linear, Polynomial, and RBF kernels is important for understanding how SVM can handle different types of data. Here's a breakdown:

1. Linear Kernel

Formula: K(xᵢ, xⱼ) = xᵢ ⋅ xⱼ (the dot product of the two data points)
Concept: This is the simplest kernel and is equivalent to using a standard linear SVM. It doesn't map the data into a higher-dimensional space; it works directly in the original feature space.
Use Case: Best suited for data that is linearly separable in its original form. It's computationally efficient and a good starting point if you suspect your data has a linear relationship.
Pros: Fast to train, easy to interpret (the hyperplane is a straight line/plane), less prone to overfitting on small datasets.
Cons: Cannot handle non-linearly separable data effectively.
2. Polynomial Kernel

Formula: K(xᵢ, xⱼ) = (γ * xᵢ ⋅ xⱼ + r)ᵈ
γ (gamma): Controls the influence of a single training example.
r (coef0): An independent term in the polynomial function.
d (degree): The degree of the polynomial.
Concept: This kernel maps the data into a higher-dimensional space by considering polynomial combinations of the original features. The decision boundary in the original space will be non-linear.
Use Case: Useful when the data has a non-linear relationship that can be captured by polynomial features. It's more flexible than the linear kernel.
Pros: Can handle some non-linear relationships.
Cons: Can become computationally expensive with a higher degree. Choosing the right degree and other parameters (γ and r) can be tricky and requires tuning. Can be prone to overfitting, especially with high degrees.


3. Radial Basis Function (RBF) Kernel

Formula: K(xᵢ, xⱼ) = exp(-γ * ||xᵢ - xⱼ||²)
γ (gamma): Controls the influence of a single training example.
Concept: This is one of the most popular and powerful kernels. It maps the data into an infinite-dimensional space. The decision boundary in the original space can be highly non-linear and complex.
Use Case: Effective for a wide range of non-linear problems where the relationship between features and the target is not easily described by a simple polynomial. It's often a good default choice when you don't have prior knowledge about the data's structure.
Pros: Can handle complex non-linear relationships effectively. Only has one main parameter to tune (γ).
Cons: Can be prone to overfitting, especially with a high γ value. Can be slower to train than the linear kernel for very large datasets. The infinite-dimensional space can be less intuitive to understand.  





#10. What is the effect of the C parameter in SVM?

Answer

  The C parameter in Support Vector Machines (SVM), particularly in Soft Margin SVM, is a crucial regularization parameter that controls the trade-off between achieving a low training error and maximizing the margin.

Here's a breakdown of its effect:

Small C:
Emphasizes a larger margin.

Allows for more misclassifications (or points within the margin) during training.

The model is more tolerant of errors and aims for a wider "street" between the classes, even if it means some data points end up on the "wrong" side.

This can lead to a simpler model and can help prevent overfitting, especially on noisy data.

However, if C is too small, the model might underfit the data, resulting in poor performance on both training and testing sets.


Large C:
Emphasizes minimizing misclassifications (or violations of the margin).

Results in a smaller margin.

The model is less tolerant of errors and tries hard to classify every training point correctly, even if it means having a narrow margin.

This can lead to a more complex model that fits the training data very closely.
However, if C is too large, the model might overfit the training data, performing very well on the training set but poorly on unseen data.
In essence, C acts as a penalty for misclassifications.

A small C means a low penalty for misclassifications, encouraging a wider margin and potentially more errors on the training set.

A large C means a high penalty for misclassifications, encouraging fewer errors on the training set but potentially a narrower margin.

Choosing the right value for C is critical for the performance of your SVM model and is typically done through hyperparameter tuning using techniques like cross-validation.

Indicates a small influence radius.

#12. What is the Naïve Bayes classifier, and why is it called "Naïve"?

answer
The Naïve Bayes classifier is a probabilistic machine learning model used for classification tasks. It's based on Bayes' Theorem, which describes the probability of an event based on prior knowledge of conditions that might be related to the event.

In simple terms, Naïve Bayes calculates the probability of a given data point belonging to a particular class, based on the probabilities of its features occurring within that class. It predicts the class with the highest probability.

It's commonly used in tasks like:

Spam filtering: Classifying emails as spam or not spam.
Text classification: Categorizing documents into different topics.
Sentiment analysis: Determining the sentiment (positive, negative, neutral) of text.
Why is it called "Naïve"?

The "Naïve" part comes from a simplifying assumption it makes: that all features are independent of each other, given the class.

In reality, this assumption is often not true. For example, in text classification, the presence of one word might be highly dependent on the presence of another word. However, despite this "naïve" assumption, the Naïve Bayes classifier often performs surprisingly well in practice, especially with large datasets.

The benefit of this independence assumption is that it makes the calculations much simpler and computationally efficient. Instead of calculating the complex joint probability of all features occurring together, it only needs to calculate the individual probabilities of each feature occurring independently within a class.

So, the name "Naïve" highlights this simplifying assumption of feature independence, which is a key characteristic of the algorithm.



#13. What is Bayes’ Theorem?


answer
  Bayes' Theorem is a fundamental concept in probability theory that describes how to update the probability of a hypothesis based on new evidence. It's named after the Reverend Thomas Bayes.

In its simplest form, Bayes' Theorem can be written as:

P(A|B) = [P(B|A) * P(A)] / P(B)

Let's break down the terms:

P(A|B): This is the posterior probability, the probability of event A happening given that event B has occurred. This is what you want to find.
P(B|A): This is the likelihood, the probability of event B happening given that event A has occurred.
P(A): This is the prior probability, the initial probability of event A happening before considering any new evidence.
P(B): This is the evidence or marginal likelihood, the probability of event B happening.
In simpler terms, Bayes' Theorem tells you how to revise your beliefs (the prior probability P(A)) in light of new evidence (event B) to get an updated belief (the posterior probability P(A|B)).

It's used in many areas, including:

Machine Learning: As we saw with Naïve Bayes, it's used for classification.
Statistics: For updating beliefs in Bayesian inference.
Medical Diagnosis: To calculate the probability of a disease given certain symptoms.
Spam Filtering: To determine the probability of an email being spam based on the words it contains.
Think of it as a way to systematically update your probability estimates as you gather more information.
   





   #14.Explain the differences between Gaussian Naïve Bayes, Multinomial Naïve Bayes, and Bernoulli Naïve Bayes.

   answer
  While all Naïve Bayes classifiers are based on Bayes' Theorem and the assumption of feature independence, they differ in how they model the distribution of the features. The choice of which one to use depends on the nature of your data.

Here are the main differences between Gaussian, Multinomial, and Bernoulli Naïve Bayes:

1. Gaussian Naïve Bayes:

Assumption: Assumes that the features follow a Gaussian (normal) distribution within each class.
How it works: It calculates the mean and standard deviation of each feature for each class. When classifying a new data point, it calculates the probability of each feature's value occurring within each class based on these Gaussian distributions.
Use Case: Typically used for continuous numerical features that are assumed to be normally distributed.
Example: Classifying emails as spam or not spam based on the length of the email, assuming the length follows a normal distribution within each class.

2. Multinomial Naïve Bayes:

Assumption: Assumes that the features represent counts or frequencies of events, such as the number of times a word appears in a document. It's based on the multinomial distribution.
How it works: It calculates the probability of each feature (e.g., each word) appearing in a document, given the class. When classifying a new document, it calculates the probability of the document belonging to each class based on the counts of the words in the document.
Use Case: Commonly used for text classification and other applications where features are counts or frequencies.
Example: Classifying news articles into different categories (sports, politics, etc.) based on the frequency of words in the articles.


3. Bernoulli Naïve Bayes:

Assumption: Assumes that the features are binary (either present or absent), represented by 0 or 1. It's based on the Bernoulli distribution.
How it works: It calculates the probability of each feature being present or absent, given the class. When classifying a new data point, it calculates the probability of it belonging to each class based on the presence or absence of the features.
Use Case: Suitable for data where features are binary or boolean, such as the presence or absence of a word in a document (as opposed to its count).
Example: Classifying documents as spam or not spam based on the presence or absence of certain keywords.

#19.Compare SVM and Naïve Bayes for classification tasks:

answer  
Laplace smoothing, also known as add-one smoothing, is a technique used in Naïve Bayes classifiers to address the zero probability problem.

Here's how it helps:

The Problem:

In Naïve Bayes, we calculate the probability of a feature value occurring within a particular class based on the counts in the training data. If a specific feature value does not appear at all in the training data for a certain class, its probability within that class will be calculated as zero.

When we multiply probabilities together in the Naïve Bayes classification formula, if any of the probabilities are zero, the entire probability for that class becomes zero, regardless of how strong the other features might be for that class. This can lead to incorrect classifications.

The Solution (Laplace Smoothing):

Laplace smoothing addresses this by adding a small constant value (usually 1) to the count of each feature value for each class, including those that had a count of zero. This constant is also added to the denominator (the total count of feature values for that class).

How it works mathematically:

Without smoothing, the probability of a feature value x_i given a class C is:

P(xᵢ | C) = (Count of xᵢ in class C) / (Total count of features in class C)

With Laplace smoothing (adding 1):

P(xᵢ | C) = (Count of xᵢ in class C + 1) / (Total count of features in class C + Number of unique features)

The effect:

By adding 1 to the count, even features that didn't appear in the training data will have a probability greater than zero.
This prevents the entire probability for a class from becoming zero due to a single unseen feature value.
It "smooths" out the probability distribution, giving a small but non-zero probability to unseen events.
In essence, Laplace smoothing is a way to avoid extreme probability estimates (zero or one) and make the Naïve Bayes model more robust to sparse data or unseen feature values in the test set. It's a simple yet effective technique to improve the performance of Naïve Bayes, especially in text classification where the vocabulary can be vast and many words might not appear in every class during training.

In Naïve Bayes, the classification decision for a new data point is based on calculating the probability of that data point belonging to each class. This calculation involves multiplying the probabilities of each of the data point's features occurring within that class.

If, during training, a particular feature value never appeared in the data for a specific class, the estimated probability of that feature value occurring given that class will be zero. When you multiply probabilities, anything multiplied by zero becomes zero. This means that even if all the other features strongly suggest that the data point belongs to that class, the presence of a single unseen feature value would make the overall probability for that class zero, preventing it from being selected as the predicted class.

Laplace smoothing prevents this by ensuring that no feature value has a zero probability within any class. It does this by adding a small count (typically 1) to the observed count of each feature value for each class, even if the observed count was originally zero. This small addition ensures that the probability calculation for any feature value, given any class, will always result in a non-zero value.

By avoiding zero probabilities, Laplace smoothing makes the Naïve Bayes model more robust and prevents it from being overly sensitive to unseen feature values in the test data. It effectively "smooths" the probability distribution, giving a small but non-zero chance to features that weren't present in the training data for a given class.

In essence, Laplace smoothing is a practical technique to make Naïve Bayes more reliable and less prone to making incorrect predictions simply because it hasn't seen a particular combination of features during training.

#37. Write a Python program to train an SVM Classifier using Linear, Polynomial, and RBF kernels on the Breast Cancer dataset and compare their accuracy

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load the Breast Cancer dataset
data = load_breast_cancer()
X, y = data.data, data.target

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# --- Train SVM with Linear kernel ---
svm_linear = SVC(kernel='linear', random_state=42)
svm_linear.fit(X_train, y_train)
y_pred_linear = svm_linear.predict(X_test)
accuracy_linear = accuracy_score(y_test, y_pred_linear)

# --- Train SVM with Polynomial kernel ---
svm_poly = SVC(kernel='poly', degree=3, random_state=42)  # degree=3 by default
svm_poly.fit(X_train, y_train)
y_pred_poly = svm_poly.predict(X_test)
accuracy_poly = accuracy_score(y_test, y_pred_poly)

# --- Train SVM with RBF kernel ---
svm_rbf = SVC(kernel='rbf', random_state=42)
svm_rbf.fit(X_train, y_train)
y_pred_rbf = svm_rbf.predict(X_test)
accuracy_rbf = accuracy_score(y_test, y_pred_rbf)

# --- Results ---
print("SVM Classifier on Breast Cancer Dataset")
print("--------------------------------------")
print(f"Accuracy with Linear Kernel:     {accuracy_linear:.4f}")
print(f"Accuracy with Polynomial Kernel: {accuracy_poly:.4f}")
print(f"Accuracy with RBF Kernel:        {accuracy_rbf:.4f}")

# Find best performing kernel
accuracies = {
    "Linear": accuracy_linear,
    "Polynomial": accuracy_poly,
    "RBF": accuracy_rbf
}
best_kernel = max(accuracies, key=accuracies.get)
print(f"\nBest performing kernel: {best_kernel}")
```

#38. Write a Python program to train an SVM Classifier using Stratified K-Fold Cross-Validation and compute the average accuracy

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.svm import SVC
import numpy as np

# Load the Breast Cancer dataset
data = load_breast_cancer()
X, y = data.data, data.target

# Define Stratified K-Fold cross-validation
k = 5  # number of folds
skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)

# Define SVM classifier (using RBF kernel as default)
svm = SVC(kernel='rbf', random_state=42)

# Perform cross-validation and compute accuracy for each fold
scores = cross_val_score(svm, X, y, cv=skf, scoring='accuracy')

# Print results
print("SVM Classifier with Stratified K-Fold Cross-Validation")
print("------------------------------------------------------")
print(f"Accuracy for each fold: {scores}")
print(f"Average accuracy: {np.mean(scores):.4f}")
```

#39. Write a Python program to train a Naïve Bayes classifier using different prior probabilities and compare performance

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Load dataset (Iris dataset for multiclass classification)
iris = load_iris()
X, y = iris.data, iris.target

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Different prior probabilities to test
priors_list = [
    None,                           # Default: priors learned from data
    [0.33, 0.33, 0.34],             # Nearly equal priors
    [0.50, 0.25, 0.25],             # Class 0 more important
    [0.10, 0.40, 0.50]              # Skewed towards class 1 and 2
]

print("Naïve Bayes Classifier with Different Priors")
print("------------------------------------------------")

# Train and evaluate for each set of priors
for priors in priors_list:
    nb = GaussianNB(priors=priors)
    nb.fit(X_train, y_train)
    y_pred = nb.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    print(f"Priors: {priors} -> Accuracy: {acc:.4f}")
```

#40. Write a Python program to perform Recursive Feature Elimination (RFE) before training an SVM Classifier and compare accuracy.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import RFE

# Load dataset
data = load_breast_cancer()
X, y = data.data, data.target

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# --- Train SVM on all features ---
svm_all = SVC(kernel="linear", random_state=42)
svm_all.fit(X_train, y_train)
y_pred_all = svm_all.predict(X_test)
accuracy_all = accuracy_score(y_test, y_pred_all)

print(f"Accuracy of SVM on all features: {accuracy_all:.4f}")

# --- Perform Recursive Feature Elimination (RFE) ---
# We'll use a linear SVM as the estimator for RFE
estimator = SVC(kernel="linear")
# Select the top 10 features
n_features_to_select = 10
rfe = RFE(estimator=estimator, n_features_to_select=n_features_to_select, step=1)
rfe.fit(X_train, y_train)

# Transform the data to include only the selected features
X_train_rfe = rfe.transform(X_train)
X_test_rfe = rfe.transform(X_test)

print(f"Original number of features: {X_train.shape[1]}")
print(f"Number of features after RFE: {X_train_rfe.shape[1]}")

# --- Train SVM on selected features ---
svm_rfe = SVC(kernel="linear", random_state=42)
svm_rfe.fit(X_train_rfe, y_train)
y_pred_rfe = svm_rfe.predict(X_test_rfe)
accuracy_rfe = accuracy_score(y_test, y_pred_rfe)

print(f"\nAccuracy of SVM on selected features (RFE): {accuracy_rfe:.4f}")

# Compare results
if accuracy_rfe > accuracy_all:
    print("\nAccuracy improved after RFE.")
elif accuracy_all > accuracy_rfe:
    print("\nAccuracy did not improve after RFE.")
else:
    print("\nAccuracy is the same with and without RFE.")

# Optional: Get the indices of the selected features
selected_feature_indices = rfe.get_support(indices=True)
print(f"\nIndices of selected features by RFE: {selected_feature_indices}")
# Optional: Get the names of selected features (if dataset has feature names)
# selected_feature_names = data.feature_names[selected_feature_indices]
# print(f"Names of selected features: {selected_feature_names}")
```

#45. Write a Python program to train a Naïve Bayes classifier and evaluate its performance using the ROC-AUC score.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import roc_curve, auc, roc_auc_score
import matplotlib.pyplot as plt

# Load the dataset
data = load_breast_cancer()
X, y = data.data, data.target

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Train a Naïve Bayes Classifier
nb = GaussianNB()
nb.fit(X_train, y_train)

# Predict probabilities for the positive class (needed for ROC-AUC)
# The second column (index 1) is the probability of the positive class
y_prob = nb.predict_proba(X_test)[:, 1]

# Compute ROC curve and ROC-AUC score
fpr, tpr, thresholds = roc_curve(y_test, y_prob)
roc_auc = auc(fpr, tpr)

# Print the ROC-AUC score
print("Naïve Bayes Classifier Evaluation using ROC-AUC")
print("------------------------------------------------")
print(f"ROC-AUC Score: {roc_auc:.4f}")

# Optional: Plot the ROC curve
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc="lower right")
plt.show()
```

#46. Write a Python program to train an SVM Classifier and visualize the Precision-Recall Curve.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import precision_recall_curve, PrecisionRecallDisplay # Import PrecisionRecallDisplay
import matplotlib.pyplot as plt

# Load the dataset
data = load_breast_cancer()
X, y = data.data, data.target

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Train an SVM Classifier (using RBF kernel by default)
# For precision-recall curve, it's better to use a model that outputs probabilities
# or a decision function. SVC with probability=True enables probability estimates.
svm_clf = SVC(kernel="rbf", random_state=42, probability=True)
svm_clf.fit(X_train, y_train)

# Get the probability of the positive class or the decision function scores
# Decision function is often preferred for SVMs when probability=True is not reliable
# For this example, we'll use predict_proba as it's common with many classifiers
y_prob = svm_clf.predict_proba(X_test)[:, 1]

# Plot the Precision-Recall Curve
prec, recall, _ = precision_recall_curve(y_test, y_prob)
display = PrecisionRecallDisplay(precision=prec, recall=recall).plot() # Use PrecisionRecallDisplay
display.ax_.set_title('Precision-Recall Curve for SVM Classifier')
plt.show()
```
